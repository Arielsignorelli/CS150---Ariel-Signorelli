# Task 1: Weekend Activities as Sets

# Create an empty set for Saturday

```python

actSat = set()

# Ask the user to enter Saturday activities
activity = input('Enter an activity from last Saturday ("Done" to stop): ')

while activity != "Done":
    actSat.add(activity)
    activity = input('Enter an activity from last Saturday ("Done" to stop): ')

# Create an empty set for Sunday
actSun = set()

# Ask the user to enter Sunday activities
activity = input('Enter an activity from last Sunday ("Done" to stop): ')

while activity != "Done":
    actSun.add(activity)
    activity = input('Enter an activity from last Sunday ("Done" to stop): ')

# Print both sets
print("Saturday activities:", actSat)
print("Sunday activities:", actSun)

# Find activities done on both days (intersection)
print("Activities done on both days:", actSat & actSun)

# Find activities done on at least one day (union)
print("Activities done on at least one day:", actSat | actSun)
```

# Task 2: Triangle Calculations

```python


import math

def triangle_solve(leg1, leg2):

    # Calculate the hypotenuse
    hypotenuse = math.sqrt(leg1**2 + leg2**2)

    # Calculate the perimeter
    perimeter = leg1 + leg2 + hypotenuse

    # Calculate the area
    area = (leg1 * leg2) / 2

    # Return the three calculations as a tuple
    return (hypotenuse, perimeter, area)


# Test the function
result = triangle_solve(3, 4)

print("Hypotenuse, Perimeter, Area:", result)
```

# Task 3: Calculator Class

```python


class Calculator:

    # Initialize the calculator attributes
    def __init__(self):
        self.prev_result = None
        self.total_calculations = 0

    # Addition method
    def add(self, a, b):
        self.prev_result = a + b
        self.total_calculations += 1
        return self.prev_result

    # Subtraction method
    def subtract(self, a, b):
        self.prev_result = a - b
        self.total_calculations += 1
        return self.prev_result

    # Multiplication method
    def multiply(self, a, b):
        self.prev_result = a * b
        self.total_calculations += 1
        return self.prev_result

    # Division method
    def divide(self, a, b):
        self.prev_result = a / b
        self.total_calculations += 1
        return self.prev_result

    # Power method
    def power(self, a, b):
        self.prev_result = a ** b
        self.total_calculations += 1
        return self.prev_result

    # Display calculator information
    def info(self):
        print("Previous Result:", self.prev_result)
        print("Total Calculations:", self.total_calculations)
```

```python

# Create a calculator object
my_calc = Calculator()

# Test addition
print("Addition:", my_calc.add(10, 5))

# Test subtraction
print("Subtraction:", my_calc.subtract(10, 5))

# Test multiplication
print("Multiplication:", my_calc.multiply(10, 5))

# Test division
print("Division:", my_calc.divide(10, 5))

# Test power
print("Power:", my_calc.power(2, 3))

# Display calculator information
my_calc.info()
```
