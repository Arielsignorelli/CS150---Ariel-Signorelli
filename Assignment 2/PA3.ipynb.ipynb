## Task 2: Student Grades and At-Risk Status

Read five student grade averages into a list, calculate the class average, minimum grade, maximum grade, and determine how many students are at risk.


```python
student_grades=[]

for i in range(5):
    grade=float(input("Enter a student grade:"))
    student_grades.append(grade)

class_average = sum(student_grades)/len(student_grades)

print("student grades:", student_grades)
print("class average:", class_average)
print("minimum grade:", min(student_grades))
print("maximun grade:", max(student_grades))

at_risk=0
for grade in student_grades:
    if grade <65:
        at_risk=at_risk +1
print(at_risk, "student are at risk in the course.") 
```

## Task 3: Ranges and Multiples

Create lists containing multiples of 2 and 3 between 1 and 100. Then create lists containing numbers that are multiples of both and numbers that are multiples of either.

```python
mult_of_2 = []

for number in range(1, 101):
    if number % 2 == 0:
        mult_of_2.append(number)

mult_of_3 = []

for number in range(1, 101):
    if number % 3 == 0:
        mult_of_3.append(number)

mult_of_both = []

for number in mult_of_2:
    if number in mult_of_3:
        mult_of_both.append(number)

mult_of_either = []

for number in range(1, 101):
    if number in mult_of_2 or number in mult_of_3:
        mult_of_either.append(number)

print("Multiples of 2:", mult_of_2)
print("Multiples of 3:", mult_of_3)
print("Multiples of both:", mult_of_both)
print("Multiples of either:", mult_of_either)
```

## Task 4: Shopping List

Create a shopping list that allows the user to enter up to seven items. The user can enter "done" to stop early. Then display the number of items, reverse order, alphabetical order, and identify duplicate items.

```python
shopping_list = []

for i in range(7):
    item = input("Enter an item to purchase or type 'done' to finish: ")

    if item.lower() == "done":
        break

    shopping_list.append(item)

print("Total number of items:", len(shopping_list))

print("Shopping list in reverse order:")
for item in reversed(shopping_list):
    print(item)

print("Shopping list in alphabetical order:")
for item in sorted(shopping_list):
    print(item)

duplicates = []

for item in shopping_list:
    count = 0

    for other_item in shopping_list:
        if item.lower() == other_item.lower():
            count = count + 1

    if count >= 2 and item not in duplicates:
        duplicates.append(item)

for item in duplicates:
    print(item, "appears at least twice in the shopping list.")
```
