## Task 3: Working with Running Times
Create varibales for two running times in minutes. Then claculate the total time, the difference between the two runs, the second run completed twice, and the average running time.

```python
run1=30
run2=25
improvement=run1-run2

# B1. Add both runs
total_time = run1 + run2
print("Total time spent running:", total_time, "minutes")

# B2. Subtract run2 from run1
improvement = run1 - run2
print("Difference between the two runs:", improvement, "minutes")

# B3. Multiply run2 by 2
twice_run2 = run2 * 2
print("Run 2 completed twice:", twice_run2, "minutes")

# B4. Calculate the average
average_time = (run1 + run2) / 2
print("Average running time:", average_time, "minutes")
```

## Task 4: String Manipulation

Create a variable called quote and assign a famous quote to it. Print the length of the quote, the quote in uppercase and lowercase, the quote with leading and trailing whitespace removed, and the quote split into a list of words.

```python
quote = "  The only way to do great work is to love what you do.  "

# B. Print the length of the quote
print("Length of quote:", len(quote))

# C. Print the quote in uppercase and lowercase
print("Uppercase:", quote.upper())
print("Lowercase:", quote.lower())

# D. Remove leading/trailing whitespace
print("Stripped quote:", quote.strip())

# E. Split the quote into a list of words
print("List of words:", quote.split())
```

## Task 5: User Input and Operations

Prompt the user to enter the lengths of the two legs of a right triangle. Use the Pythagorean theorem to calculate the hypotenuse and print the answer rounded to the nearest one-hundredth.

The formula is:

c = √(a² + b²)

```python
a = float(input("Enter the length of the first leg: "))
b = float(input("Enter the length of the second leg: "))

c = (a**2 + b**2)**(1/2)

print("The hypotenuse of your triangle is", round(c, 2), "units.")
```
