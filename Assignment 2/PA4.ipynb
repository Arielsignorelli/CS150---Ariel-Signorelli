## Task 1: Input Positive Integer and Factors

### Part A
Ask the user for a positive integer and print all positive factors of that integer.

### Part B
Determine whether the user's positive integer is prime.

# Task 1A: Find all factors

```python


number = int(input("Please input a positive integer: "))

while number <= 0:
    number = int(input("Please input a positive integer: "))

for factor in range(1, number + 1):
    if number % factor == 0:
        print(factor, "is a factor of", number)


# Task 1B: Determine whether the number is prime

is_prime = True

if number == 1:
    is_prime = False
else:
    for factor in range(2, number):
        if number % factor == 0:
            is_prime = False
            break

if is_prime:
    print("Your number is prime.")
else:
    print("Your number is not prime.")
```


## Task 2: Language Translator

Create an English-to-Irish translation dictionary containing ten words. Continue asking for words until the user enters an empty string.


```python


translations = {
    "hello": "dia duit",
    "goodbye": "slán",
    "cat": "cat",
    "dog": "madra",
    "chair": "cathaoir",
    "book": "leabhar",
    "water": "uisce",
    "house": "teach",
    "bread": "arán",
    "thank you": "go raibh maith agat"
}

word = input(
    "Enter a word to translate from English to Irish "
    "(hit enter to stop): "
).strip().lower()

while word != "":
    if word in translations:
        print("The Irish translation for", word, "is",
              translations[word].upper() + ".")
    else:
        print("The Irish translation for", word,
              "is not in this dictionary.")

    word = input(
        "Enter a word to translate from English to Irish "
        "(hit enter to stop): "
    ).strip().lower()

print("Translation program ended.")
```

# Task 3: Louisville monthly temperatures in Fahrenheit
# Each list contains [average high, average low]

```python


temperatures = {
    "january": [43, 27],
    "february": [49, 30],
    "march": [58, 37],
    "april": [69, 46],
    "may": [77, 56],
    "june": [85, 65],
    "july": [88, 69],
    "august": [87, 67],
    "september": [81, 60],
    "october": [70, 48],
    "november": [57, 38],
    "december": [46, 30]
}

month = input(
    "Enter a month to see Louisville's average high and low temperatures (or quit): "
).strip().lower()

while month != "quit":
    if month in temperatures:
        high = temperatures[month][0]
        low = temperatures[month][1]

        print("The average high temperature in Louisville for",
              month.capitalize(), "is", high,
              "and the average low is", str(low) + ".")
    else:
        print("Not a valid month.")

    month = input(
        "Enter a month to see Louisville's average high and low temperatures (or quit): "
    ).strip().lower()

print("Temperature program ended.")
```


## Task 4: Inventory Dictionary and Prices

Create a hardware store inventory dictionary. Allow the user to look up an item, find items under a budget, calculate total quantity, or calculate the total value of the inventory.


```python



inventory = {
    "hammer": [15, 12.49],
    "wrench": [20, 7.99],
    "screwdriver": [40, 3.99],
    "nailgun": [10, 36.59],
    "padlock": [25, 5.99],
    "roll_of_duct_tape": [30, 4.95]
}

print("Hardware Store Inventory Menu")
print("1. Look up an item")
print("2. Find items below a budget")
print("3. Calculate total quantity of items")
print("4. Calculate total value of inventory")

choice = input("Enter your choice (1-4): ")

if choice == "1":
    item = input("Enter an item name: ").strip().lower()

    if item in inventory:
        quantity = inventory[item][0]
        price = inventory[item][1]

        print("Quantity available:", quantity)
        print("Price per item: $" + format(price, ".2f"))
    else:
        print("This item is not available.")

elif choice == "2":
    budget = float(input("Enter your maximum budget: $"))

    for item in inventory:
        price = inventory[item][1]

        if price < budget:
            print(item, "- $" + format(price, ".2f"))

elif choice == "3":
    total_quantity = 0

    for item in inventory:
        total_quantity = total_quantity + inventory[item][0]

    print("Total quantity of items:", total_quantity)

elif choice == "4":
    total_value = 0

    for item in inventory:
        quantity = inventory[item][0]
        price = inventory[item][1]

        total_value = total_value + quantity * price

    print("Total inventory value: $" + format(total_value, ".2f"))

else:
    print("Invalid menu choice.")
```
